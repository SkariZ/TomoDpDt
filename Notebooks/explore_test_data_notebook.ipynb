## Exploring Available Test Data

### **1. Introduction**

This notebook provides an overview of the **precomputed synthetic test objects** available in the `test_data/` folder.
These volumes serve as standardized test cases for validating different parts of the tomographic reconstruction pipeline — including forward simulation, rotation recovery, and optimization-based reconstruction.

Each dataset represents a different type of 3D object (e.g., smooth Gaussian fields, layered structures, phantoms) and can be used directly without any additional preprocessing.

### **2. Load and Inspect Available Datasets**


```python
data_folder = "../test_data"
available_datasets = sorted([f for f in os.listdir(data_folder) if f.endswith(".npy") and 'vol' in f])

print(f"Found {len(available_datasets)} test datasets:\n")
for name in available_datasets:
    print(" -", name)
```

### **3. Load and Visualize a Test Object**
